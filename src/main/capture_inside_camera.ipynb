```python
import cv2
import time
import datetime
from pathlib import Path
```

```python
# Open the default camera
cam = cv2.VideoCapture('rtsp://admin:Bolotova123@192.168.88.254:554/Streaming/Channels/301', cv2.CAP_FFMPEG)
```

```python
# Get the default frame width and height
frame_width = int(cam.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cam.get(cv2.CAP_PROP_FRAME_HEIGHT))
```

```python
# Define the codec and create VideoWriter object
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
```

```python
import numpy as np

def is_gray_frame(frame, threshold=10):
    # Проверяем, является ли кадр "серым" (мало деталей)
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    return cv2.meanStdDev(gray)[1][0][0] < threshold
```

```python
frame_count = 0
save_interval = 1  # Интервал сохранения (в секундах)

last_time = time.time()

while True:
    ret, frame = cam.read()
    if not ret:
        break
    if not ret or is_gray_frame(frame):
        continue  # Пропускаем серый/пустой кадр

    current_time = time.time()
    if current_time - last_time >= save_interval:
        filename = f"frame_{datetime.datetime.fromtimestamp(current_time).strftime("%Y-%m-%d_%a_%H-%M-%S")}.jpg"
        cv2.imwrite(Path("./my_camera_inside") / filename, frame)
        print(f"Сохранён {filename}")
        frame_count += 1
        last_time = current_time

    # Выход по нажатию 'q' (если нужно, но окна нет)
    if cv2.waitKey(1) == ord('q'):
        break

cam.release()
```
